# Comparaison expérimentale des méthodes de prédiction

Ce notebook compare différentes méthodes de répartition des élèves
dans les maisons de Poudlard.

L’indicateur utilisé est le nombre d’erreurs calculé avec la fonction
`nb_erreurs`, en comparant les prédictions avec l’affectation réelle
du choixpeau.


```python
from methode_score import *
from methode_distance import *
import json
```

```python
# Réponses des élèves
answers = create_answers_from_texte_file("questionnaire_premiere_annee_10q.txt")

# Affectation réelle du choixpeau
f = open("affectation_premiere_annee.json", "r")
real_affectation = json.load(f)
f.close()

# Références (4 fondateurs)
f = open("houses_ref.json", "r")
refs_4 = json.load(f)
f.close()

# Références (40 magiciens)
f = open("houses_multiple_refs.json", "r")
refs_40 = json.load(f)
f.close()
```

# Méthode avec les scores

```python
a = lecture_reponses("questionnaire_premiere_annee.txt")
b = repartition(a)
err_m_score = nb_erreurs(b, real_affectation)
print("Erreur methode_score :" , err_m_score)
```

# Distance euclidienne avec 4 références

```python
repartition_eucl_4 = Euclidean_repartition(answers, refs_4)
err_eucl_4 = nb_erreurs(repartition_eucl_4, real_affectation)

print("Erreur euclidienne (4 références) :", err_eucl_4)
```

# Distance euclidienne avec 40 références

```python
repartition_eucl_40 = Euclidean_repartition(answers, refs_40)
err_eucl_40 = nb_erreurs(repartition_eucl_40, real_affectation)

print("Erreur euclidienne (40 références) :", err_eucl_40)
```

# Méthode des k plus proches voisins

```python
errors_knn = {}

for k in range(1, 6):
    repartition_knn = NN_repartition(answers, refs_40, k)
    errors_knn[k] = nb_erreurs(repartition_knn, real_affectation)
    print("Erreur k-NN (k =", k, ") :", errors_knn[k])
```

# Tableau récapitulatif

```python
print("\nRécapitulatif des erreurs :")
print(" Erreur methode_score:", err_m_score)
print("Euclidienne (4 refs) :", err_eucl_4)
print("Euclidienne (40 refs) :", err_eucl_40)

for k in errors_knn:
    print("k-NN (k =", k, ") :", errors_knn[k])
```

# Conclusion

La méthode basée sur la distance euclidienne améliore les résultats par
rapport à la méthode basée sur les scores.

L’utilisation de 40 références permet une diminution supplémentaire du
nombre d’erreurs, ce qui montre que des références plus nombreuses
représentent mieux chaque maison.

La méthode des k plus proches voisins est la plus performante.
On observe qu’une valeur intermédiaire de k donne les meilleurs résultats,
car un k trop petit est sensible au bruit et un k trop grand lisse
excessivement les décisions.

La meilleure méthode est donc celle des k plus proches voisins avec la
valeur de k minimisant le nombre d’erreurs.(donc ici k = 3)
